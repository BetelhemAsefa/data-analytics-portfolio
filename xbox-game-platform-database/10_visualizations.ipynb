```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sp_developer_game_revenue_visualization.csv")

# Sort by revenue
df_sorted = df.sort_values(by="item_revenue", ascending=False)

# Cumulative %
cumulative = df_sorted["item_revenue"].cumsum() / df_sorted["item_revenue"].sum() * 100

fig, ax1 = plt.subplots(figsize=(12, 6))

# Bars
ax1.bar(df_sorted["title"], df_sorted["item_revenue"], color='skyblue')
ax1.set_ylabel("Revenue ($)", color='blue')
ax1.tick_params(axis='x', rotation=90)

# Line (cumulative %)
ax2 = ax1.twinx()
ax2.plot(df_sorted["title"], cumulative, color='red', marker='o')
ax2.set_ylabel("Cumulative %", color='red')
ax2.set_ylim([0, 110])

# 80% reference line
ax2.axhline(80, color='green', linestyle='--')

plt.title("Pareto Chart of Item Revenue")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sp_get_player_spend_summary_visualization.csv")
df.head()
```

```python
plt.figure(figsize=(12, 6))

# Bottom bar: number of orders
plt.bar(
    df["p_player_id"].astype(str),
    df["@p_order_count"],
    label="Order Count"
)

# Top part of stack: total spend
plt.bar(
    df["p_player_id"].astype(str),
    df["@p_total_spend"],
    bottom=df["@p_order_count"],
    label="Total Spend ($)"
)

plt.xlabel("Player ID")
plt.ylabel("Value (Order Count + Spend)")
plt.title("Stacked Bar Chart of Player Orders and Spending")
plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load file
df = pd.read_csv("purchase_auditing_visualization.csv")

# Select numeric columns only
numeric_df = df[["audit_id", "purchase_id", "player_id", "item_id", "amount"]]

# Compute correlation matrix
corr = numeric_df.corr()

# Plot heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr, annot=True, cmap="Blues", fmt=".2f", linewidths=0.5)

plt.title("Correlation Heatmap – Purchase Auditing Table")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("sp_purchases_visualization.csv")

# Convert the date column properly
df["purchased_at"] = pd.to_datetime(df["purchased_at"])

df_daily = df.groupby(df["purchased_at"].dt.date)["amount"].sum()
```

```python
plt.figure(figsize=(12, 6))

plt.plot(df_daily.index, df_daily.values, marker="o", linewidth=2)

plt.xlabel("Date")
plt.ylabel("Total Spending ($)")
plt.title("Daily Purchase Revenue Trend")
plt.grid(True, linestyle="--", alpha=0.6)
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("summary_game_daily_visualization.csv")

df["summary_date"] = pd.to_datetime(df["summary_date"])

df["cohort_date"] = df.groupby("game_id")["summary_date"].transform("min")

df["cohort_index"] = (df["summary_date"] - df["cohort_date"]).dt.days
```

```python
cohort_table = df.pivot_table(
    index="cohort_date",
    columns="cohort_index",
    values="total_revenue",
    aggfunc="sum"
).fillna(0)
```

```python
plt.figure(figsize=(12, 6))
sns.heatmap(cohort_table, annot=True, fmt=".2f", cmap="Blues")

plt.title("Cohort Chart – Revenue Growth by Game Cohort")
plt.xlabel("Cohort Age (Days)")
plt.ylabel("Cohort Start Date")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load data
df = pd.read_csv("summary_player_total_visualization.csv")

# Define funnel stages
stage_labels = [
    "Total Players",
    "Purchased ≥ 1",
    "Purchased ≥ 2",
    "High Spenders"
]

# Values for each stage
total_players = df.shape[0]
purchased_once = df[df["total_purchases"] >= 1].shape[0]
purchased_twice = df[df["total_purchases"] >= 2].shape[0]

# Define high-spender threshold
threshold = df["total_spend"].median()  # or choose a specific value like 60
high_spenders = df[df["total_spend"] >= threshold].shape[0]

stage_values = [
    total_players,
    purchased_once,
    purchased_twice,
    high_spenders
]

# Plot funnel
plt.figure(figsize=(8, 6))
plt.plot(stage_values, stage_labels, marker="o", linewidth=4)
plt.gca().invert_yaxis()
# Funnel shape (largest on top)

plt.title("Player Lifecycle Funnel Chart")
plt.xlabel("Number of Players")
plt.grid(True, linestyle="--", alpha=0.6)

plt.tight_layout()
plt.show()
```
